## GRÁFICOS DE BARRAS - 3M

### PUNTOS

```python
top3 = dfutil[['Club','Pts']].head(3)
media_pts = dfutil['Pts'].mean()
df = pd.DataFrame([['Media', media_pts]],columns=['Club', 'Pts'])
df_pts = pd.concat([top3,df])

# Configuración del estilo
sns.set_style("white")
plt.figure(figsize=(6, 6))

# Crear gráfico de barras
barplot = sns.barplot(x='Club', y='Pts', data=df_pts,palette=['navy', 'blue', 'dodgerblue', 'red'], saturation=0.8, width = 0.8)

# Personalización
plt.xlabel('')
plt.xticks(rotation=0, fontsize = 12)
sns.despine(left=True, bottom=False, right=True, top=True)
# Añadir etiquetas de valor
for p in barplot.patches:
    barplot.annotate(format(p.get_height(), '.1f'), (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom', xytext=(0, 5), textcoords='offset points', fontsize = 20)

# Ajustar layout
plt.tight_layout()
plt.show()
```
